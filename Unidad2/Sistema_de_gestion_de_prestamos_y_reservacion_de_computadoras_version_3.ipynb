```python
import datetime
from datetime import date, datetime
alumnos=[
    ["26040200", "Dimetrio" , "Ingenieria en sistemas"],
    ["26040201", "Manbatman", "Fisica cuantica"],
    ["26040202", "Zero"     , "Robotica"],
    ["26040203", "Shadow"   , "Ingenieria en bioquimica"],
    ["26040204", "", ""]
    ]
computadoras=[
             ["PC-LC1-01", "LC1", "Python, VS code",       "Disponible"   ],
             ["PC-LC1-02", "LC1", "Python, Java, VS code", "Prestada"     ],
             ["PC-LC1-03", "LC1", "Autocad, Mathlab",      "Reservada"    ],
             ["PC-LC2-01", "LC2", "Python, VS code",       "Disponible"   ],
             ["PC-LC2-02", "LC2", "Python, Java, VS code", "Prestada"     ],
             ["PC-LC2-03", "LC2", "Autocad, Mathlab",      "Reservada"    ],
             ["PC-LC3-01", "LC3", "Python, VS code",       "Disponible"   ],
             ["PC-LC3-02", "LC3", "Python, Java, VS code", "Prestada"     ],
             ["PC-LC3-03", "LC3", "Autocad, Mathlab",      "Reservada"    ],
             ["PC-LC3-04", "LC3", "Python, VS code",       "Mantenimiento"]
             ]
prestamos=[
    [alumnos[0][0], computadoras[1][0], computadoras[1][2], computadoras[1][3]],
    [alumnos[1][0], computadoras[4][0], computadoras[4][2], computadoras[4][3]],
    [alumnos[2][0], computadoras[7][0], computadoras[7][2], computadoras[7][3]],
    ]
requerimiento=[
]
reservaciones=[
    [alumnos[0], computadoras[2], ""]
]

def listar_computadoras():
  print("\n---Computadoras---")
  for i in range(len(computadoras)):
    print("Aulas: ", computadoras[i][1], "  ID: ", computadoras[i][0], "  Sofware: ", computadoras[i][2], "  Estado: ", computadoras[i][3])
    print("-------------------------------------------------------------------------------------------------")
  print(prestamos)

def listar_sofware():
  print("\n---Sofware---")
  programa=input("Dime el programa que necesites en la computadora ")
  for i in range(len(computadoras)):
      if(programa.lower() in computadoras[i][2].lower()):
        print("  Sofware: ", computadoras[i][2], "  ID: ", computadoras[i][0], "Aulas: ", computadoras[i][1], "  Estado: ", computadoras[i][3])

def prestamo():
  print("\n---Solicitar prestamo---")
  num_control=input("Dame tu numero de control ")
  for i in range(len(alumnos)):
    if num_control in alumnos[i]:
      programa=input("Dime el programa que necesites en la computadora ")
      for i in range(len(computadoras)):
        if(programa.lower() in computadoras[i][2].lower()):
          if(computadoras[i][3]=="Disponible"):
            print("Se asigno la computadora", computadoras[i][0])
            computadoras[i][3]="Prestada"
            prestamos.append([num_control, computadoras[i][0], computadoras[i][2], computadoras[i][3],])
            break
        else:
          if (i+1) == len(computadoras):
            print("No hay computadoras disponibles con el sofware requerido")
            j=0
            while j!=1 and j!=2:
              print("Deseas pedir que se agrege este sofware a alguna computadora?")
              agregar_sofware=input("1:Si 2:No ")
              if agregar_sofware=="1":
                requerimiento.append([num_control, programa, datetime.now()])
                break
              elif agregar_sofware==2:
                break
    else:
      print("numero de control incorrecto, escribe uno correcto")

def devolucion():
  print("\n-------------Devolver computadora-------------")
  pc= input("Dame el id de la computadora ")
  existe = False
  for i in range(len(computadoras)):
    if computadoras[i][0] == pc:
      existe=True
      if computadoras[i][3] == "Prestada" or computadoras[i][3] == "Reservada":
        computadoras[i][3] = "Disponible"
        print("La computadora", computadoras[i][0], "ya esta disponible")
        break
      else:
        print("La computadora no esta ocupada")
  if(existe==False):
    print("No existe la computadora '", pc, "'")

def agregar():
  pc=input("Dime el id de la nueva computadora")
  aula=input("Dime el aula de la nueva computadora")
  sofware=input("Dime los sofware de la nueva computadora")
  for i in range(len(computadoras)):
    if pc==computadoras[i][0]:
      print("No se puede agregar la computadora porque ya existe")
      break
    else:
      if (i+1) == len(computadoras):
        computadoras.append([pc, aula, sofware, "Disponible"])
        print("La computadora se agrego con exito")

def lista():
  for i in range(len(requerimiento)):
    print("El alumno con numero de control ", requerimiento[i][0], " pidio el sofware ", requerimiento[i][1], " el dia y hora ", requerimiento[i][2])

def reservacion():
  print("")

opcion=0
while opcion<=7:
  print("+=======================================+")
  print("|          Sistema de prestamos         |")
  print("+=======================================+")
  print("| 1.Consultar computadoras              |")
  print("| 2.Buscar por sofware                  |")
  print("| 3.Solicitar prestamo                  |")
  print("| 4.Devolver computadora                |")
  print("| 5.Crear reservacion                   |")
  print("| 6.Agregar computadora                 |")
  print("| 7.Revisar lista de sofware requeridos |")
  print("| 8.Salir                               |")
  print("+=======================================+")
  opcion=int(input("Seleccione una opcion "))
  if(opcion==1):
    listar_computadoras()
  elif(opcion==2):
    listar_sofware()
  elif(opcion==3):
    prestamo()
  elif(opcion==4):
    devolucion()
  elif(opcion==5):
    print("opcion 5")
  elif(opcion==6):
    agregar()
  elif(opcion==7):
    lista()
  elif(opcion==8):
    print("Hasta la proxima")
```
