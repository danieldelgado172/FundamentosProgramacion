```python
import time
alumnos=["26040200", "26040201", "26040202", "26040203", "26040204"]
computadoras=[
             ["PC-LC1-01", "LC1", "Python, VS code",       "Disponible"],
             ["PC-LC1-02", "LC1", "Python, Java, VS code", "Prestada"  ],
             ["PC-LC1-03", "LC1", "Autocad, Mathlab",      "Reservada" ],
             ["PC-LC2-01", "LC2", "Python, VS code",       "Disponible"],
             ["PC-LC2-02", "LC2", "Python, Java, VS code", "Prestada"  ],
             ["PC-LC2-03", "LC2", "Autocad, Mathlab",      "Reservada" ],
             ["PC-LC3-01", "LC3", "Python, VS code",       "Disponible"],
             ["PC-LC3-02", "LC3", "Python, Java, VS code", "Prestada"  ],
             ["PC-LC3-03", "LC3", "Autocad, Mathlab",      "Reservada" ],
             ["PC-LC3-04", "LC3", "Python, VS code",       "Mantenimiento"]
             ]
def listar_computadoras():
  print("\n---Computadoras---")
  for i in range(len(computadoras)):
    print("Aulas: ", computadoras[i][1], "  ID: ", computadoras[i][0], "  Sofware: ", computadoras[i][2], "  Estado: ", computadoras[i][3])
    print("-------------------------------------------------------------------------------------------------")

def listar_sofware():
  print("\n---Sofware---")
  programa=input("Dime el programa que necesites en la computadora ")
  for i in range(len(computadoras)):
      if(programa.lower() in computadoras[i][2].lower()):
        print("tlabaja")
  ##Escribir solo los que tengan el sofware requerido

def prestamo():
  print("\n---Solicitar prestamo---")
  num_control=input("Dame tu numero de control ")
  if num_control in alumnos:
    programa=input("Dime el programa que necesites en la computadora ")
    for i in range(len(computadoras)):
      if(programa.lower() in computadoras[i][2].lower()):
        if(computadoras[i][3]=="Disponible"):
          print("Se asigno la computadora", computadoras[i][0])
          computadoras[i][3]="Prestada"
          break
      else:
        if (i+1) == len(computadoras):
          print("No hay computadoras disponibles con el sofware requerido")
  else:
    print("numero de control incorrecto, escribe uno correcto")

def devolucion():
  print("\n-------------Devolver computadora-------------")
  pc= input("Dame el id de la computadora ")
  existe = False
  for i in range(len(computadoras)):
    if computadoras[i][0] == pc:
      existe=True
      if computadoras[i][3] == "Prestada" or computadoras[i][3] == "Reservada":
        computadoras[i][3] = "Disponible"
        print("La computadora", computadoras[i][0], "ya esta disponible")
        break
      else:
        print("La computadora no esta ocupada")
  if(existe==False):
    print("No existe la computadora '", pc, "'")

def agregar():
  print("trabajando en esta opcion...")
opcion=0
while opcion<=6:
  print("+==========================+")
  print("|   Sistema de prestamos   |")
  print("+==========================+")
  print("| 1.Consultar computadoras |")
  print("| 2.Buscar por sofware     |")
  print("| 3.Solicitar prestamo     |")
  print("| 4.Devolver computadora   |")
  print("| 5.Crear reservacion      |")
  print("| 6.Agregar computadora    |")
  print("| 7.Salir                  |")
  print("+==========================+")
  opcion=int(input("Seleccione una opcion "))
  if(opcion==1):
    listar_computadoras()
  elif(opcion==2):
    listar_sofware()
  elif(opcion==3):
    prestamo()
  elif(opcion==4):
    devolucion()
  elif(opcion==5):
    print("opcion 5")
  elif(opcion==6):
    agregar()
  elif(opcion==7):
    print("Hasta la proxima")
```
