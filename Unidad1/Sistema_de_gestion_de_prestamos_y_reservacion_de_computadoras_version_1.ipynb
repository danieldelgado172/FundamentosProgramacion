```python
alumnos=["26040200", "26040201", "26040202", "26040203", "26040204"]
computadoras=["PC-LC1-01", "PC-LC1-02", "PC-LC1-03",
              "PC-LC2-01", "PC-LC2-02", "PC-LC2-03",
              "PC-LC3-01", "PC-LC3-02", "PC-LC3-03", "PC-LC3-04"]
aulas=["LC1", "LC1", "LC1",
       "LC2", "LC2", "LC2",
       "LC3", "LC3", "LC3", "LC3",]
sofware=["Python, VS code", "Python, Java, VS code", "Autocad, Mathlab",
         "Python, VS code", "Python, Java, VS code", "Autocad, Mathlab",
         "Python, VS code", "Python, Java, VS code", "Autocad, Mathlab", "Python, VS code", ]
Estados=["Disponible", "Prestada", "Reservada",
         "Disponible", "Prestada", "Reservada",
         "Disponible", "Prestada", "Reservada", "Mantenimiento"]
opcion=0
while opcion<=6:
  print("+==========================+")
  print("|   Sistema de prestamos   |")
  print("+==========================+")
  print("| 1.Consultar computadoras |")
  print("| 2.Buscar por sofware     |")
  print("| 3.Solicitar prestamo     |")
  print("| 4.Devolver computadora   |")
  print("| 5.Crear reservacion      |")
  print("| 6.Agregar computadora    |")
  print("| 7.Salir                  |")
  print("+==========================+")
  opcion=int(input("Seleccione una opcion "))
  if(opcion==1):
    print("\n---Computadoras---")
    for i in range(len(computadoras)):
      print("Aulas: ", aulas[i], "  ID: ", computadoras[i], "  Sofware: ", sofware[i], "  Estado: ", Estados[i])
      print("-------------------------------------------------------------------------------------------------")
  elif(opcion==2):
    print("\n---Programas---")
    for i in range(len(sofware)):
      print("Sofware: ", sofware[i], "  ID: ", computadoras[i], " Aulas: ", aulas[i], "  Estado: ", Estados[i])
      print("-------------------------------------------------------------------------------------------------")
  elif(opcion==3):
    print("\n---Solicitar prestamo---")
    num_control=input("Dame tu numero de control ")
    if num_control in alumnos:
      programa=input("Dime el programa que necesites en la computadora ")
      for i in range(len(computadoras)):
        if(programa.lower() in sofware[i].lower()):
          if(Estados[i]=="Disponible"):
            print("Se asigno la computadora", computadoras[i])
            Estados[i]="Prestada"
            break
          elif (i+1) == len(computadoras):
            print("No hay computadoras disponibles")
    else:
      print("numero de control incorrecto, escribe uno correcto")
  elif(opcion==4):
    print("\n-------------Devolver computadora-------------")
    pc= input("Dame el id de la computadora ")
    existe = False

    for i in range(len(computadoras)):
      if computadoras[i] == pc:
        existe=True
        if Estados[i] == "Prestada" or Estados[i] == "Reservada":
          Estados[i] = "Disponible"
          print("La computadora ya esta disponible")
          break
        else:
          print("La computadora no esta ocupada")
    if(existe==False):
      print("No existe la computadora '", pc, "'")
  elif(opcion==5):
    print("opcion 5")
  elif(opcion==6):
    print("tlabaja")
  elif(opcion==7):
    print("Hasta la proxima")
```
